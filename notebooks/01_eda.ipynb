# CompLens: survey exploration
Run from the repository root after `python -m ml.data.generate`. Production EDA lives in `ml/evaluation/eda.py`. All salaries are INR/year.

```python
import pandas as pd
from ml.common import RAW
df = pd.read_csv(RAW)
display(df.describe(include='all').T)
display((df.isna().mean() * 100).rename('missing_pct'))
```

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df.plot.scatter(x='experience_years', y='salary', alpha=.12, s=8, ax=axes[0])
axes[0].set(xlabel='Experience (years)', ylabel='Annual salary (INR/year)')
df.salary.hist(bins=40, ax=axes[1])
axes[1].set(xlabel='Annual salary (INR/year)', ylabel='Candidates (count)')
plt.tight_layout()
```

```python
for feature in ['education_level','job_role','industry','city']:
    display(df.groupby(feature).salary.agg(['count','median']).sort_values('median', ascending=False))
```
